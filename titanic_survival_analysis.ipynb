```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# load dataset
df = pd.read_csv("Titanic-Dataset.csv")

# display first five records
df.head()
```

```python

print("dataset shape:", df.shape)


print("\ncolumns:")
print(df.columns.tolist())


print("\ndataset information:")
df.info()


df.describe()
```

```python
print(df.isnull().sum())
```

```python

survival_counts = df["Survived"].value_counts()

print(survival_counts)



plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Survived")

plt.title("Passenger Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
survival_percentage = df["Survived"].value_counts(normalize=True) * 100

print(survival_percentage)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Sex", hue="Survived")

plt.title("Survival Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
gender_survival = df.groupby("Sex")["Survived"].mean() * 100

print(gender_survival)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Pclass", hue="Survived")

plt.title("Survival Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
class_survival = df.groupby("Pclass")["Survived"].mean() * 100

print(class_survival)
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(data=df, x="Age", bins=30, kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(data=df, x="Survived", y="Age")

plt.title("Age Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.show()
```

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

print(df.groupby("FamilySize")["Survived"].mean() * 100)

plt.figure(figsize=(9, 5))

sns.barplot(data=df, x="FamilySize", y="Survived")

plt.title("Survival Rate by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Survival Rate")

plt.show()
```

```python
family_analysis = df.groupby("FamilySize")["Survived"].agg(
    ["count", "sum", "mean"]
)

family_analysis["SurvivalPercentage"] = family_analysis["mean"] * 100

family_analysis
```

## 2. Exploratory Data Analysis — Key Findings

### 2.1 Overall Survival
- Total passengers: 891
- Passengers who survived: 342 (38.38%)
- Passengers who did not survive: 549 (61.62%)

### 2.2 Gender Analysis
- Female survival rate: 74.20%
- Male survival rate: 18.89%
- Gender appears to be a strong predictor of survival.

### 2.3 Passenger Class Analysis
- First-class survival rate: 62.96%
- Second-class survival rate: 47.28%
- Third-class survival rate: 24.24%
- Passenger class shows a clear relationship with survival.

### 2.4 Age Analysis
- Most passengers were between 20 and 35 years old.
- The age distribution is right-skewed.
- Median ages of survivors and non-survivors are relatively similar.
- Age may provide additional predictive information.

### 2.5 Family Size Analysis
- Family size was derived using SibSp + Parch + 1.
- Passengers travelling in smaller families generally showed higher survival rates.
- Larger family groups generally showed lower survival rates.
- Small group sizes should be considered before interpreting these results.

### Overall Conclusion
Gender and passenger class show particularly strong relationships with survival. Age and family size may also contribute to predictive performance.

These findings will guide feature selection and preprocessing in the next stages of the project.
